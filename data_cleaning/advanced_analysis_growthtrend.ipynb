# Load clear data from charts

```python
### Section: Compute 2014–2023 Growth for ALL Indicators (wide format CSV)
import pandas as pd

# Load CSV
df = pd.read_csv("../charts/RAA04_final_2014_2023.csv")

# Identify metadata & indicator columns
meta_cols = ["Year", "NUTS 2 Region"]
indicator_cols = [c for c in df.columns if c not in meta_cols]

# Prepare result table
growth_rows = []

regions = df["NUTS 2 Region"].unique()

for region in regions:
    d14 = df[(df["Year"] == 2014) & (df["NUTS 2 Region"] == region)]
    d23 = df[(df["Year"] == 2023) & (df["NUTS 2 Region"] == region)]

    if len(d14) == 0 or len(d23) == 0:
        continue

    row = {"Region": region}

    for col in indicator_cols:
        v14 = d14[col].values[0]
        v23 = d23[col].values[0]

        # handle missing data
        if pd.isna(v14) or pd.isna(v23):
            row[col + " (Growth %)"] = None
        else:
            pct = (v23 - v14) / v14 * 100
            row[col + " (Growth %)"] = pct

    row["Period"] = "2014–2023"
    growth_rows.append(row)

growth_df = pd.DataFrame(growth_rows)

# Optional: save to CSV for Vega-Lite lookup
growth_df.to_csv("../charts/NUTS2_growth_all_indicators_2014_2023.csv", index=False)

growth_df
```
